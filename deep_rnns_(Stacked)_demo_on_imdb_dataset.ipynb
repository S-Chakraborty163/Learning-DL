```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense,LSTM,GRU
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)
```

```python
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
model_RNN = Sequential([
    Embedding(10000, 32, input_length=100),
    SimpleRNN(5, return_sequences=True),   # RNN layer with 5 units
    SimpleRNN(5),                          # Another RNN layer with 5 units
    Dense(1, activation='sigmoid')
])
```

```python
model_LSTM = Sequential([
    Embedding(10000, 32, input_length=100),
    LSTM(5, return_sequences=True),
    LSTM(5),
    Dense(1, activation='sigmoid')
])
```

```python
model_GRU = Sequential([
    Embedding(10000, 32, input_length=100),
    GRU(5, return_sequences=True),
    GRU(5),
    Dense(1, activation='sigmoid')
])
```

**Stacked RNN:**

```python
model_RNN.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history_RNN = model_RNN.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
```

**Stacked LSTM:**

```python
model_LSTM.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history_LSTM = model_LSTM.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
```

**Stacked GRU:**

```python
model_GRU.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history_GRU = model_GRU.fit(x_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
```
